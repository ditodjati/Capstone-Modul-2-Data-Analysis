# **Background**

TransJakarta adalah sistem transportasi Bus Rapid Transit (BRT) pertama di Asia Tenggara yang mulai beroperasi sejak tahun 2004. Layanan ini dirancang untuk menjadi solusi transportasi publik massal yang cepat, efisien, dan terjangkau, khususnya untuk warga DKI Jakarta dan sekitarnya. Berbeda dengan moda transportasi umum lain, TransJakarta memiliki jalur khusus atau busway yang dirancang untuk meminimalisir hambatan lalu lintas, sehingga perjalanan dapat berlangsung lebih lancar dan tepat waktu.

Seiring berjalannya waktu, TransJakarta berkembang menjadi salah satu tulang punggung mobilitas masyarakat urban. Saat ini, sistem TransJakarta mencakup ratusan rute yang tersebar dalam berbagai koridor utama maupun rute pengumpan (feeder), menjangkau berbagai wilayah strategis di ibu kota. Jaringan rute yang luas ini memungkinkan penumpang untuk berpindah dari satu wilayah ke wilayah lain dengan mudah dan relatif cepat.

TransJakarta juga mengadopsi teknologi modern dalam sistem pembayarannya. Penumpang diwajibkan untuk melakukan **tap-in** saat memasuki halte dan **tap-out** saat keluar. Proses ini menggunakan kartu elektronik (e-money) yang diterbitkan oleh berbagai bank seperti Flazz (BCA), E-Money (Mandiri), Brizzi (BRI), dan JakCard (Bank DKI). Sistem ini tidak hanya meningkatkan efisiensi, tetapi juga memungkinkan pencatatan perjalanan secara digital yang dapat dimanfaatkan untuk keperluan analisis dan pengambilan keputusan.

Sebagai layanan transportasi yang terus berkembang dan digunakan oleh jutaan pengguna setiap bulan, TransJakarta memiliki potensi besar untuk ditingkatkan dari sisi efisiensi dan pengalaman pengguna melalui pemanfaatan data digital yang dihasilkan setiap harinya.

Data cleaning dilakukan untuk memastikan bahwa dataset yang akan dianalisis berada dalam kondisi optimal. Tahapan yang dilakukan mencakup:

1. Konversi kolom waktu ke format datetime
2. Pembuatan kolom durasi perjalanan

### 1. Konversi Kolom Waktu & Pembuatan Kolom Durasi

```python
# Konversi waktu tap-in dan tap-out
df['tapInTime'] = pd.to_datetime(df['tapInTime'], errors='coerce')
df['tapOutTime'] = pd.to_datetime(df['tapOutTime'], errors='coerce')

# Buat kolom durasi perjalanan (dalam menit)
df['duration_min'] = (df['tapOutTime'] - df['tapInTime']).dt.total_seconds() / 60
```

**Insight:**  
**Halte Penjaringan, Garuda Taman Mini, dan BKN** adalah 3 titik masuk penumpang terbanyak — cocok dijadikan prioritas untuk manajemen antrean dan tambahan armada.

---

```python
# Tambahkan kolom hari
df['tapInDay'] = pd.to_datetime(df['tapInTime']).dt.day_name()

# Urutkan hari sesuai kalender
hari_urut = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
plt.figure(figsize=(10, 5))
sns.countplot(x='tapInDay', data=df, order=hari_urut, palette="viridis")
plt.title("Distribusi Penumpang Berdasarkan Hari")
plt.xlabel("Hari")
plt.ylabel("Jumlah Penumpang")
plt.show()
```

**Insight:**  
**Jumlah transaksi tertinggi terjadi pada hari kerja (Senin–Jumat)** dengan total ±6.800 per hari, dibandingkan hanya ±1.800-an di akhir pekan.

---

```python
# Hitung frekuensi penumpang per hari
hari_counts = df['tapInDay'].value_counts().reindex(hari_urut)

# Hitung expected value jika distribusi merata
expected = [hari_counts.sum() / 7] * 7

# Uji chi-square
chi2, p = chisquare(f_obs=hari_counts, f_exp=expected)

print("Chi-Square:", chi2)
print("p-value:", p)

if p < 0.05:
    print("Distribusi penumpang tidak merata secara signifikan antar hari.")
else:
    print("Distribusi penumpang cenderung merata antar hari.")
```

### Pengujian Statistik: Distribusi Penumpang Berdasarkan Hari

Untuk mengetahui apakah penumpang TransJakarta tersebar secara merata sepanjang hari, dilakukan uji **Chi-Square Goodness of Fit** terhadap distribusi frekuensi tap-in per hari.

Hasil uji menunjukkan bahwa nilai **p < 0.05**, sehingga dapat disimpulkan bahwa distribusi jumlah penumpang **tidak merata secara signifikan antar hari**. Ini menguatkan temuan visual bahwa hari kerja (weekday) lebih sibuk dibanding akhir pekan.

**Insight:** TransJakarta dapat fokus mengalokasikan armada tambahan pada hari kerja, terutama di koridor-koridor padat.

**Insight:**  
Sebagian besar koridor menunjukkan variasi durasi perjalanan yang cukup tinggi, dengan rentang antar kuartil (IQR) sekitar 40–50 menit. Misalnya, koridor **Kebayoran Lama – Tanah Abang** memiliki IQR yang lebar, menunjukkan ketidakkonsistenan waktu tempuh yang cukup signifikan. Selain itu, beberapa koridor seperti **Kalideres – Bundaran HI via Veteran** bahkan memiliki pencilan (outliers) ekstrem, yang mengindikasikan adanya perjalanan dengan durasi jauh di atas rata-rata. Variasi ini dapat disebabkan oleh kemacetan, waktu tempuh yang tidak stabil, atau faktor operasional lainnya.

---

```python
# Buat kolom jam
df['tapInHour'] = pd.to_datetime(df['tapInTime']).dt.hour

# Pisahkan jam sibuk dan non-sibuk
peak_hours = [6, 7, 8, 16, 17, 18]
df_peak = df[df['tapInHour'].isin(peak_hours) & df['duration_min'].notnull()]
df_non_peak = df[~df['tapInHour'].isin(peak_hours) & df['duration_min'].notnull()]

# Uji t-test
t_stat, p_val = ttest_ind(df_peak['duration_min'], df_non_peak['duration_min'], equal_var=False)

print("T-Statistic:", t_stat)
print("p-value:", p_val)

if p_val < 0.05:
    print("Durasi perjalanan pada jam sibuk berbeda secara signifikan dibandingkan jam non-sibuk.")
else:
    print("Tidak terdapat perbedaan signifikan durasi antara jam sibuk dan non-sibuk.")
```

### Pengujian Statistik: Durasi Perjalanan di Jam Sibuk vs Non-Sibuk

Untuk menguji apakah perjalanan TransJakarta cenderung lebih lama di jam sibuk, dilakukan uji statistik **Two-Sample T-Test** antara durasi rata-rata di jam sibuk (06.00–09.00 & 16.00–19.00) dan jam non-sibuk.

Hasil uji menunjukkan nilai **p < 0.05**, artinya terdapat **perbedaan yang signifikan secara statistik**. Durasi perjalanan di jam sibuk memang lebih lama, kemungkinan besar karena kemacetan dan kepadatan penumpang.

**Insight bisnis:** Bukti ini mendukung pentingnya penguatan aturan jalur khusus busway dan penambahan armada di jam sibuk.
